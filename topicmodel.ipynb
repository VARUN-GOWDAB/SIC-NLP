```python
import pandas as pd
import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
documents = [
    "I love programming in Python.",
    "Python is a great programming language.",
    "I enjoy solving problems with Python.",
    "Python is versatile and powerful.",
    "I dislike bugs in my code.",
    "Debugging can be frustrating.",
    "I find programming challenging but rewarding.",
    "I prefer Python over other programming languages."
]

vectorizer=TfidfVectorizer()
```

```python
x=vectorizer.fit_transform(documents)
```

```python
print("Feature names:", vectorizer.get_feature_names_out())
```

```python
print(x.shape) 
```

```python
from sklearn.decomposition import TruncatedSVD

svd = TruncatedSVD(n_components=2, random_state=42)
svd.fit(x)

print(svd.components_)
```

```python
for i in enumerate(svd.components_):
    idx, component = i
    print('Component', idx + 1)
    for index in component.argsort()[-5:][::-1]:
        print(vectorizer.get_feature_names_out()[index], component[index])
```

```python
svd.singular_values_
```

```python
document_topics=pd.DataFrame(svd.transform(x), columns=['Topic 1', 'Topic 2'])
```

```python
import numpy as np

# =====================================================
# 1. VOCABULARY
# =====================================================

words = ["I", "eat", "an", "apple", "every", "morning"]

word_to_index = {
    "I": 0,
    "eat": 1,
    "an": 2,
    "apple": 3,
    "every": 4,
    "morning": 5
}

V = 6              # Vocabulary size
D = 3              # Embedding dimension


# =====================================================
# 2. INITIALIZE WEIGHTS
# =====================================================

np.random.seed(42)

# W: Input → Hidden
# Shape = 3 × 6
W = np.random.randn(D, V) * 0.01

# W': Hidden → Output
# Shape = 6 × 3
W_prime = np.random.randn(V, D) * 0.01


# =====================================================
# 3. ONE-HOT FUNCTION
# =====================================================

def one_hot(index, size):
    vector = np.zeros(size)
    vector[index] = 1
    return vector


# =====================================================
# 4. SOFTMAX FUNCTION
# =====================================================

def softmax(scores):

    # Prevent numerical overflow
    scores = scores - np.max(scores)

    exp_scores = np.exp(scores)

    return exp_scores / np.sum(exp_scores)


# =====================================================
# 5. TRAINING DATA
# =====================================================

# Context words
context_words = ["an", "every"]

# Target word
target_word = "apple"

# Convert target into index
target_index = word_to_index[target_word]


# =====================================================
# 6. TRAINING
# =====================================================

learning_rate = 0.1
epochs = 1000


for epoch in range(epochs):

    # -------------------------------------------------
    # FORWARD PASS
    # -------------------------------------------------

    # One-hot vectors
    v_an = one_hot(word_to_index["an"], V)
    v_every = one_hot(word_to_index["every"], V)


    # -------------------------------------------------
    # Input → Hidden
    # -------------------------------------------------

    h_an = np.dot(W, v_an)

    h_every = np.dot(W, v_every)


    # -------------------------------------------------
    # Average context embeddings
    # -------------------------------------------------

    x = (h_an + h_every) / 2


    # -------------------------------------------------
    # Hidden → Output
    # -------------------------------------------------

    scores = np.dot(W_prime, x)


    # -------------------------------------------------
    # Softmax
    # -------------------------------------------------

    probabilities = softmax(scores)


    # -------------------------------------------------
    # LOSS
    # -------------------------------------------------

    loss = -np.log(probabilities[target_index])


    # =================================================
    # BACKPROPAGATION
    # =================================================

    # Copy probabilities
    error = probabilities.copy()

    # For correct word:
    # probability - 1
    error[target_index] -= 1


    # -------------------------------------------------
    # Gradient for W'
    # -------------------------------------------------

    # dW' = error × hidden_vector
    dW_prime = np.outer(error, x)


    # -------------------------------------------------
    # Gradient for hidden vector
    # -------------------------------------------------

    dx = np.dot(W_prime.T, error)


    # -------------------------------------------------
    # Because x is average of two embeddings:
    #
    # x = (h_an + h_every) / 2
    #
    # Each embedding receives half of the gradient
    # -------------------------------------------------

    dh_an = dx / 2
    dh_every = dx / 2


    # -------------------------------------------------
    # Update W
    # -------------------------------------------------

    # W columns corresponding to the context words
    W[:, word_to_index["an"]] -= learning_rate * dh_an

    W[:, word_to_index["every"]] -= learning_rate * dh_every


    # -------------------------------------------------
    # Update W'
    # -------------------------------------------------

    W_prime -= learning_rate * dW_prime


    # -------------------------------------------------
    # Print progress
    # -------------------------------------------------

    if epoch % 100 == 0:

        predicted_index = np.argmax(probabilities)
        predicted_word = words[predicted_index]

        print(
            "Epoch:", epoch,
            "Loss:", round(loss, 4),
            "Prediction:", predicted_word
        )


# =====================================================
# 7. FINAL PREDICTION
# =====================================================

print("\n==============================")
print("FINAL RESULT")
print("==============================")

print("Context:", context_words)
print("Target:", target_word)

print("\nProbabilities:")

for word, probability in zip(words, probabilities):
    print(word, ":", round(probability, 4))

predicted_index = np.argmax(probabilities)
predicted_word = words[predicted_index]

print("\nPredicted word:", predicted_word)


# =====================================================
# 8. LEARNED EMBEDDINGS
# =====================================================

print("\nLearned Embedding Matrix W:")
print(W)
```
