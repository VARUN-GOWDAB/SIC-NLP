```python
from sklearn.feature_extraction.text import CountVectorizer
#BOW
```

```python
documents = [
    "learning intelligence machine learning statistics",
    "machine classification learning performance",
    "machine classification, machine learning, machine performance"
]
```

```python
vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)
```

```python
vectorizer.get_feature_names_out()
```

```python
X.toarray() #DTM
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer()

X_tfidf = tfidf.fit_transform(documents)
```

```python
tfidf.get_feature_names_out()
```

```python
X_tfidf.toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(X_tfidf)

similarity
```

```python
reviews = [
    "The product is really good and I absolutely love it!",
    "This product is terrible. I don't recommend it.",
    "The delivery was very fast and the packaging was excellent.",
    "I am disappointed because the product stopped working after two days.",
    "Amazing quality! The product works perfectly.",
    "The item was okay, but the delivery was extremely slow.",
    "I loved the design, but the product is not durable.",
    "Very poor quality and terrible customer service.",
    "The product is affordable, useful, and easy to use.",
    "I am not happy with this purchase."
]
```

```python
vectorizer = CountVectorizer()

X = vectorizer.fit_transform(reviews)
```

```python
vectorizer.get_feature_names_out()
```

```python
X.toarray() #DTM
```

```python
tfidf = TfidfVectorizer()

X_tfidf = tfidf.fit_transform(reviews)
```

```python
tfidf.get_feature_names_out()
```

```python
X_tfidf.toarray()
```

```python
similarity = cosine_similarity(X_tfidf)

similarity
```
