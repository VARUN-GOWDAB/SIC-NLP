```python
import json
```

In json format the keys or the string values must be in double quotes

```python
varun_json = '''{
    "name": "varun",
    "age": 25,
    "sem": 5,
    "skills": ["python", "java", "c++"]
}'''
```

```python
varun_dct=json.loads(varun_json)
```

```python
print(varun_dct)
```

```python
v_json=json.dumps(varun_dct,indent=4)
```

```python
print(type(v_json))
```

```python
print(v_json)
```

```python
with open('dboss.json','w') as f:
    json.dump(varun_dct,f,ensure_ascii=False,indent=4)
    
```
